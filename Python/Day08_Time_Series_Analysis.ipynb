#  Day 08 - Time Series Analysis

##  Objective

Analyze sales and profit trends over time to understand business performance and identify seasonal patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("C:/Data Projects/Ecommerce_Sales_Analysis/Data/Superstore.xlsx")
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
```

```python
df[["Order Date","Year","Month","Month Name"]].head()
```

## Sales by Year

```python
sales_by_year = (
    df.groupby("Year")["Sales"]
      .sum()
)

sales_by_year
```

```python
sales_by_year.plot(kind="line", marker="o", figsize=(8,5))

plt.title("Total Sales by Year")
plt.xlabel("Year")
plt.ylabel("Sales")

plt.show()
```

## Sales by Month

```python
month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

sales_by_month = (
    df.groupby("Month Name")["Sales"]
      .sum()
      .reindex(month_order)
)

sales_by_month
```

```python
sales_by_month.plot(kind="bar", figsize=(10,5))

plt.title("Sales by Month")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```

## Profit by Year

```python
profit_by_year = (
    df.groupby("Year")["Profit"]
      .sum()
)

profit_by_year
```

```python
profit_by_year.plot(kind="line", marker="o", figsize=(8,5))

plt.title("Profit by Year")
plt.xlabel("Year")
plt.ylabel("Profit")

plt.show()
```

## Key Findings

### Sales by Year

- Sales increased consistently from 2014 to 2017.
- 2017 recorded the highest annual sales.

### Profit by Year

- Profit increased every year.
- 2017 achieved the highest annual profit.

### Sales by Month

- November and December generated the highest sales.
- January and February recorded the lowest sales.
- Sales appear to increase during the last quarter of the year, suggesting possible seasonal demand.
