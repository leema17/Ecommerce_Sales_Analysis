## Day 05 - Exploratory Data Analysis (EDA)

## Objective
Calculate key business metrics and understand overall business performance.

```python
import pandas as pd

df = pd.read_excel("C:/Data Projects/Ecommerce_Sales_Analysis/Data/Superstore.xlsx")
```

### Calculated Total Sales.

```python
total_sales = df["Sales"].sum()

print(f"Total Sales: {total_sales:,.2f}")
```

### Calculated Total Profit.

```python
total_profit = df["Profit"].sum()

print(f"Total Profit: {total_profit:,.2f}")
```

### Counted the total number of unique orders.

```python
total_orders = df["Order ID"].nunique()

print(f"Total Orders: {total_orders}")
```

### Counted the total number of unique customers.

```python
total_customers = df["Customer ID"].nunique()

print(f"Total Customers: {total_customers}")
```

### Calculated the Average Order Value (AOV).

```python
order_sales = df.groupby("Order ID")["Sales"].sum()

average_order_value = order_sales.mean()

print(f"Average Order Value: {average_order_value:,.2f}")
```

## Profit Margin

```python
profit_margin = (total_profit / total_sales) * 100

print(f"Profit Margin: {profit_margin:.2f}%")
```

### Calculated the Average Orders per Customer.

```python
average_orders = total_orders / total_customers

print(f"Average Orders per Customer: {average_orders:.2f}")
```
