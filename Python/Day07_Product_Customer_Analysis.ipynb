#  Day 07 - Product & Customer Analysis

##  Objective

Analyze products and customers to identify top performers and products that generate losses.

```python
import pandas as pd

df = pd.read_excel("C:/Data Projects/Ecommerce_Sales_Analysis/Data/Superstore.xlsx")
```

## Top 10 Products by Sales

```python
top_sales_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_sales_products
```

## Top 10 Products by Profit

```python
top_profit_products = (
    df.groupby("Product Name")["Profit"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_profit_products
```

## Bottom 10 Products by Profit

```python
bottom_profit_products = (
    df.groupby("Product Name")["Profit"]
      .sum()
      .sort_values()
      .head(10)
)

bottom_profit_products
```

## Top 10 Customers by Sales

```python
top_customers = (
    df.groupby("Customer Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```

## Top Customers With High Sales and Profit 

```python
customer_analysis = (
    df.groupby("Customer Name")
      .agg({
          "Sales": "sum",
          "Profit": "sum"
      })
      .sort_values(by="Sales", ascending=False)
)

customer_analysis.head(10)
```

```python
customer_analysis["Profit Margin %"] = (
    customer_analysis["Profit"] / customer_analysis["Sales"]
) * 100

customer_analysis.sort_values(by="Sales", ascending=False).head(10)
```

## Top Customers With High Profit Margin

```python
customer_analysis.sort_values(
    by="Profit Margin %",
    ascending=False
).head(10)
```
